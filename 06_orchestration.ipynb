# Task 6 — Orchestrate with a Databricks Job

To accomplish this orchestration, create a Databricks Job with four sequential tasks: bronze ingestion -> silver transform -> gold model -> data quality checks.

### Key Steps and Requirements:
1. **Job Parameter:** Pass the target catalog name as a job parameter. Each notebook can read this using `dbutils.widgets.get` or SQL parameters at runtime.
2. **Task Dependencies:** Configure task dependencies to enforce that no task starts before its predecessor completes successfully. This is done in the Jobs UI or with the task graph in a Job JSON spec.
3. **Retries:** Set retry counts for each task to tolerate transient failures (typically 1–3 retries).
4. **Failure Handling:** Make the data quality check task critical; if it fails, configure the job to fail entirely (set as dependent task and disable downstream execution on failure).
5. **Successful Run:** After all tasks complete, verify job status and download or screenshot the run summary for documentation.
6. **Production Scheduling & Monitoring:**
   - Schedule the pipeline using the Jobs UI (cron or periodic options).
   - Monitor runs in the Lakeflow Jobs dashboard; view logs, alerts, and Data Quality Monitoring results.
