```python
dbutils.widgets.text("catalog", "dev")
catalog = dbutils.widgets.get("catalog")
```

```python
from pyspark.sql import functions as F

df = spark.read.table("samples.tpch.customer")

df = df.withColumns(
    {
        "file_path": F.col("_metadata.file_path"),
        "ingestion_time": F.current_timestamp(),
    }
)

df.write.mode("overwrite").option(
    "comment", "Bronze layer: customer data ingested from samples.tpch.customer with source file path and ingestion timestamp."
).saveAsTable(f"{catalog}.bronze.customer")
```
